```python
water_quality_data['ammonia'] = water_quality_data['ammonia'].apply(lambda x:float(x))
```

```python
water_quality_data['is_safe'] = water_quality_data['is_safe'].apply(lambda x:int(x))
```

```python
plt.figure(figsize=(20,8))
sns.boxplot(water_quality_data)
plt.show()
```

```python
num_cols = water_quality_data.select_dtypes(include=['int64','float64'])

q1 = num_cols.quantile(0.25)
q3 = num_cols.quantile(0.75)

iqr = q3 - q1

outlier_data = ((num_cols < (q1 - 1.5*iqr)) |
                (num_cols > (q3 + 1.5*iqr)))

outlier_data.sum()
```

```python
for i in water_quality_data.columns:
    print(i)
    print("Skew = {}, kurtosis = {}".format(water_quality_data[i].skew(), water_quality_data[i].kurtosis()))
    sns.histplot(water_quality_data[i], kde=True)
    plt.show()
```

```python
from sklearn.preprocessing import PowerTransformer
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

# Split
x = water_quality_data.drop(columns=['is_safe'])
y = water_quality_data['is_safe']

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, stratify=y, random_state=42
)

# Pipeline
pipeline_randomForest = Pipeline([
    ('transform', PowerTransformer(method='yeo-johnson')),
    ('smote', SMOTE()),
    ('model', RandomForestClassifier())
])
# Train
pipeline_randomForest.fit(x_train, y_train)
# Evaluate
print("Random Forest")
print("Train Score:", pipeline_randomForest.score(x_train, y_train))
print("Test Score:", pipeline_randomForest.score(x_test, y_test))

pipeline_decisionTree = Pipeline([
    ('transform', PowerTransformer(method='yeo-johnson')),
    ('smote', SMOTE()),
    ('model', DecisionTreeClassifier())
])
pipeline_decisionTree.fit(x_train, y_train)
# Evaluate
print("Decision Tree")
print("Train Score:", pipeline_decisionTree.score(x_train, y_train))
print("Test Score:", pipeline_decisionTree.score(x_test, y_test))

pipeline_logisticReg = Pipeline([
    ('transform', PowerTransformer(method='yeo-johnson')),
    ('smote', SMOTE()),
    ('model', LogisticRegression())
])
pipeline_logisticReg.fit(x_train, y_train)
# Evaluate
print("Logistic reg")
print("Train Score:", pipeline_logisticReg.score(x_train, y_train))
print("Test Score:", pipeline_logisticReg.score(x_test, y_test))

pipeline_svc = Pipeline([
    ('transform', PowerTransformer(method='yeo-johnson')),
    ('smote', SMOTE()),
    ('model', SVC(probability=True))
])
pipeline_svc.fit(x_train, y_train)
# Evaluate
print("SVC")
print("Train Score:", pipeline_svc.score(x_train, y_train))
print("Test Score:", pipeline_svc.score(x_test, y_test))
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, roc_auc_score
```

```python
# Random Forest
from sklearn.model_selection import GridSearchCV, StratifiedKFold

random_forest_param_grid = {
    'model__n_estimators': [100, 200],
    'model__max_depth': [5, 10]
}

cv = StratifiedKFold(n_splits=5)

random_forest_grid = GridSearchCV(
    pipeline_randomForest,
    random_forest_param_grid,
    cv=cv,
    scoring='f1'
)

random_forest_grid.fit(x_train, y_train)

print("Best Params:", random_forest_grid.best_params_)
print("Best Score:", random_forest_grid.best_score_)

predict_random_forest = random_forest_grid.predict(x_test)
print("Random Forest ")
print("accuracy_score = ", accuracy_score(y_test, predict_random_forest))
print("F1 score=", f1_score(y_test, predict_random_forest))
print("roc_auc_score=", roc_auc_score(y_test, predict_random_forest))
```

```python
# Logisitc
pipeline_logistic = Pipeline([
    ('transform', PowerTransformer(method='yeo-johnson')),
    ('smote', SMOTE()),
    ('model', LogisticRegression())
])

lr_param_grid = {
    'model__C': [0.1, 1, 10],
    'model__penalty': ['l2'],
    'model__solver': ['lbfgs'],
    'model__max_iter': [100, 200]
}

cv = StratifiedKFold(n_splits=5)

logistic_grid = GridSearchCV(
    estimator=pipeline_logistic,
    param_grid=lr_param_grid,
    cv=cv,
    scoring='f1',
    n_jobs=-1
)

logistic_grid.fit(x_train, y_train)

print("Best Params:", logistic_grid.best_params_)
print("Best CV Score:", logistic_grid.best_score_)

# Evaluation
y_pred = logistic_grid.predict(x_test)
y_proba = logistic_grid.predict_proba(x_test)[:, 1]

print("Logistic Regression")
print("Accuracy =", accuracy_score(y_test, y_pred))
print("F1 Score =", f1_score(y_test, y_pred))
print("ROC-AUC =", roc_auc_score(y_test, y_proba))
```

```python
# Decision tree

dtree_param_grid = {
    'model__max_depth': [5, 10]
}

cv = StratifiedKFold(n_splits=5)

decision_tree_grid = GridSearchCV(
    pipeline_decisionTree,
    dtree_param_grid,
    cv=cv,
    scoring='f1'
)

decision_tree_grid.fit(x_train, y_train)

print("Best Params:", decision_tree_grid.best_params_)
print("Best Score:", decision_tree_grid.best_score_)

predict_decision_tree = decision_tree_grid.predict(x_test)
predict_prob_decision_tree = decision_tree_grid.predict_proba(x_test)[:,1]
print("Decision Tree ")
print("accuracy_score = ", accuracy_score(y_test, predict_decision_tree))
print("F1 score=", f1_score(y_test, predict_decision_tree))
print("roc_auc_score=", roc_auc_score(y_test, predict_prob_decision_tree))
```

```python
param_grid_svc = {
    'model__C': [0.1, 1, 10],
    'model__kernel': ['linear', 'rbf'],
    'model__gamma': ['scale', 'auto']
}

svc_grid = GridSearchCV(
    estimator=pipeline_svc,
    param_grid=param_grid_svc,
    cv=5,
    scoring='f1',
    n_jobs=-1
)

svc_grid.fit(x_train, y_train)

print("Best Params:", svc_grid.best_params_)
print("Best CV Score:", svc_grid.best_score_)

# Evaluation
y_pred = svc_grid.predict(x_test)
y_proba = svc_grid.predict_proba(x_test)[:, 1]

print("SVC")
print("Accuracy =", accuracy_score(y_test, y_pred))
print("F1 Score =", f1_score(y_test, y_pred))
print("ROC-AUC =", roc_auc_score(y_test, y_proba))
```
